```python
import pandas as pd # Import the pandas library for data manipulation and analysis, aliased as 'pd'
import numpy as np # Import the numpy library for numerical operations and missing value handling, aliased as 'np'
```

```python
df = pd.read_csv('Nigerian_Movies') # Read the CSV file named 'Nigerian_Movies' into a DataFrame called 'df'
df.head()  # Display the first 5 rows of the DataFrame to inspect the initial data structure
df.head()
```

```python
fake_plot = "Add a plot in your language" # Define a placeholder string used by IMDb for missing plot descriptions
df['show_desc'] = df['show_desc'].replace(fake_plot, np.nan)# Replace all instances of the placeholder in the 'show_desc' column with NaN (missing value)
```

```python
df = df.drop(
    columns=["total_votes"]
)  # Remove the 'total_votes' column from the DataFrame as it is not needed
df = df.dropna(
    subset=["title"]
)  # Remove any rows where the 'title' column contains a missing (NaN) value

df["genre"] = df["genre"].fillna(
    "Unknown"
)  # Replace missing values in the 'genre' column with the text 'Unknown'
df["director"] = df["director"].fillna(
    "Unknown"
)  # Replace missing values in the 'director' column with the text 'Unknown'
df["stars"] = df["stars"].fillna(
    "Unknown"
)  # Replace missing values in the 'stars' column with the text 'Unknown'
df["show_desc"] = df["show_desc"].fillna(
    ""
)  # Replace missing values in the 'show_desc' column with an empty string

df = df.reset_index(
    drop=True
)  # Reset the DataFrame index to be continuous (0, 1, 2...) after dropping rows, discarding the old index
```

```python
def recommend_by_genre(
    genre_input, num_recommendations=10
):  # Define a function that accepts a target genre string and optional max recommendation count (default 10)
    genre_input = (
        genre_input.strip().lower()
    )  # Remove leading/trailing whitespace and convert the input genre string to lowercase
    mask = (
        df["genre"].str.lower().str.contains(genre_input, na=False)
    )  # Create a boolean series checking if 'genre_input' is contained within each movie's genre string (case-insensitive)
    matches = df[
        mask
    ].copy()  # Filter the DataFrame using the boolean mask and create an independent copy of matching rows
    matches = matches.sort_values(
        by="rating", ascending=False, na_position="last"
    )  # Sort the matching movies by 'rating' from highest to lowest, placing unrated (NaN) movies at the end
    results = matches[
        ["title", "year", "genre", "director", "stars", "rating", "show_desc"]
    ]  # Select only the relevant display columns from the DataFrame
    return results.head(
        num_recommendations
    )  # Return the top N recommendations based on 'num_recommendations'
```

```python
import os  # Module for interacting with operating system environment variables
import requests  # Library for sending HTTP requests to the TMDB API
from dotenv import (
    load_dotenv,
)  # Import load_dotenv to read key-value pairs from your .env file

load_dotenv()  # Load environment variables defined in the local .env file into os.environ
TMDB_API_KEY = os.getenv(
    "TMDB_API_KEY"
)  # Safely retrieve the TMDB API key stored under the variable name 'TMDB_API_KEY'


def search_poster(
    title, year=None
):  # Define function to search for a movie poster by title and optional year
    url = "https://api.themoviedb.org/3/search/movie"  # Base endpoint URL for TMDB movie search API
    params = {
        "api_key": TMDB_API_KEY,  # Pass the securely loaded API key instead of hardcoding it
        "query": title,  # Pass the movie title as the search query
    }

    if year and pd.notna(
        year
    ):  # Check if year is provided and is not NaN/null in pandas
        params["year"] = int(
            year
        )  # Convert release year to integer and append to request parameters

    response = requests.get(
        url, params=params
    )  # Send HTTP GET request to TMDB API with query parameters
    data = (
        response.json()
    )  # Parse JSON response body into a Python dictionary object

    if data.get("results"):  # Check if search returned any matching movies
        poster_path = data["results"][0].get(
            "poster_path"
        )  # Extract relative poster path from the top search result
        if poster_path:  # Verify that a valid poster path string exists
            return f"https://image.tmdb.org/t/p/w500{poster_path}"  # Return full TMDB poster URL with 500px resolution

    return None  # Return None if no matching movie or poster path was found
```

```python
import os  # Import operating system module to read environment variables
import requests  # Import requests to handle API calls
from dotenv import (
    load_dotenv,
)  # Import load_dotenv to read variables from .env file

load_dotenv()  # Load environment variables from .env into os.environ
TMDB_API_KEY = os.getenv(
    "TMDB_API_KEY"
)  # Retrieve the TMDB API key safely from environment variables


def search_poster(
    title, year=None
):  # Define function to search for a poster with optional year matching
    url = "https://api.themoviedb.org/3/search/movie"  # Base URL endpoint for TMDB movie search API
    params = {
        "api_key": TMDB_API_KEY,  # Pass the securely loaded API key instead of hardcoding it
        "query": title,  # Search query parameter set to the movie title
    }

    response = requests.get(
        url, params=params
    )  # Send HTTP GET request to TMDB API with search parameters
    data = response.json()  # Parse JSON response into a Python dictionary
    results = data.get(
        "results"
    )  # Extract the list of movie results from the response data

    if not results:  # Check if the results list is empty or None
        return None  # Return None immediately if no matching movies were found

    if year and pd.notna(
        year
    ):  # Check if a target year was provided and is not NaN
        target_year = int(year)  # Convert the target year to an integer
        best_match = None  # Initialize variable to track the best matching movie result
        best_diff = None  # Initialize variable to track the smallest year difference

        for movie in results:  # Iterate through each movie in the search results
            release_date = movie.get(
                "release_date"
            )  # Extract the release date string (e.g. '2022-12-16')
            if not release_date:  # Check if release date is missing or empty
                continue  # Skip to next movie if release date isn't present

            try:
                movie_year = int(
                    release_date[:4]
                )  # Extract the first 4 characters (YYYY) and convert to integer
            except ValueError:  # Handle cases where year parsing fails
                continue  # Skip to next movie if release date year format is invalid

            diff = abs(
                movie_year - target_year
            )  # Calculate absolute difference between result year and target year

            if (
                best_diff is None or diff < best_diff
            ):  # Check if this is the first valid result or closer than previous best
                best_diff = diff  # Update best_diff with the smaller year difference
                best_match = (
                    movie  # Update best_match with the current movie object
                )

        if (
            best_match and best_diff is not None and best_diff <= 1
        ):  # Check if a match was found within 1 year tolerance
            poster_path = best_match.get(
                "poster_path"
            )  # Retrieve poster path relative URL from the best match
            if poster_path:  # Check if the poster path exists
                return f"https://image.tmdb.org/t/p/w500{poster_path}"  # Return full TMDB poster image URL (500px width)
            return None  # Return None if the best matching movie has no poster

        return None  # Return None if no movie matched within the 1-year tolerance window

    poster_path = results[0].get(
        "poster_path"
    )  # Fallback: if no target year was provided, get poster path from first result
    if poster_path:  # Check if poster path exists for the first result
        return f"https://image.tmdb.org/t/p/w500{poster_path}"  # Return full TMDB poster URL for the first result

    return None  # Return None if the first result has no poster path available
```
